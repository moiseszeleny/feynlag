## 1. The model

Three Higgs doublets, all with $(SU(2)_L,U(1)_Y)=(\mathbf 2,\tfrac12)$. Under S₃ the pair
$(H_1,H_2)$ is a **doublet** and $H_S$ a **singlet**. Writing
$x_{ij}\equiv H_i^\dagger H_j$ for $i,j\in\{1,2,S\}$ — so $x_{S1}=H_S^\dagger H_1$,
$x_{S2}=H_S^\dagger H_2$, and $x_{iS}=x_{Si}^*$ — the S₃ Clebsch–Gordan decomposition
$\mathbf2\otimes\mathbf2=\mathbf1\oplus\mathbf1'\oplus\mathbf2$ in feynlag's real-orthogonal
irrep basis gives

$$\underbrace{x_{11}+x_{22}}_{\mathbf 1},\qquad
\underbrace{x_{12}-x_{21}}_{\mathbf 1'},\qquad
\underbrace{d_2=\big(x_{11}-x_{22},\,-(x_{12}+x_{21})\big)}_{\mathbf 2},\qquad
x_S\equiv(x_{S1},x_{S2}),$$

and the most general renormalizable S₃- and gauge-invariant potential is

$$
V=\mu_1^2\,(x_{11}+x_{22})+\mu_0^2\,x_{SS}
+\lambda_1(x_{11}+x_{22})^2+\lambda_2(x_{12}-x_{21})^2+\lambda_3\,|d_2|^2
$$
$$
+\;\lambda_4\big[x_{S1} d_2^{(1)}+x_{S2} d_2^{(2)}+\text{h.c.}\big]
+\lambda_5\,x_{SS}(x_{11}+x_{22})
+\lambda_6\big[x_{S1} x_{1S}+x_{S2} x_{2S}\big]
+\lambda_7\big[x_{S1}^2+x_{S2}^2+\text{h.c.}\big]
+\lambda_8\,x_{SS}^2 .
$$

($x_{S1}x_{1S}+x_{S2}x_{2S}=|x_{S1}|^2+|x_{S2}|^2$ since $x_{iS}=x_{Si}^*$ — manifestly
real, as a coupling in a potential must be.)

This is `examples/thdm_s3.py`'s potential, rebuilt by `model.build_model`. Two other
parametrizations of the *same* potential appear in the literature — [DasDey14] Eq. (3c)
also uses $\lambda_1\ldots\lambda_8$, and [GomezBock21] Eq. (2)/(6) uses $a,\ldots,h$.
**They are not the same $\lambda$'s**; §4 settles the dictionary.

First: is the potential actually invariant? `build_model(check=True)` runs feynlag's gauge,
S₃, hermiticity and mass-dimension checks.


```python
t0 = time.time()
m = build_model(check=True)          # raises if any invariance check fails
report = m.model.check_invariance()
print("built and checked in %.1f s" % (time.time() - t0))
print(report, "  failures:", report.failures)

assert report.ok and report.checked > 0
ok("potential passes all %d invariance checks (gauge, S3, hermiticity, mass dimension)"
   % report.checked)
```

A check that cannot fail proves nothing. The same machinery must **reject** a term S₃
forbids — here $x_{SS}\,(x_{11}-x_{22})$, which is the first component of a doublet, not a
singlet:


```python
from feynlag import check_discrete_invariance, dag

H1, H2, HS = m.doublets
bra = lambda A, B: (dag(A) * B.mat)[0]
x11, x22, x12, x21 = bra(H1, H1), bra(H2, H2), bra(H1, H2), bra(H2, H1)
xSS = bra(HS, HS)

allowed, _ = check_discrete_invariance(xSS * (x11 + x22), m.s3)   # the lambda5 invariant
forbidden, _ = check_discrete_invariance(xSS * (x11 - x22), m.s3) # a doublet component

print("S3-invariant   x_SS (x11 + x22) :", allowed)
print("S3-invariant   x_SS (x11 - x22) :", forbidden)
assert allowed and not forbidden
ok("the invariance check discriminates: it accepts the singlet and rejects the doublet")
```

## 2. The vacuum

### 2.1 The √3 alignment is forced, not chosen

Each neutral component gets a VEV, $H_i^0\to(v_i+h_i+ia_i)/\sqrt2$, and the three tadpole
conditions $\partial V/\partial v_i=0$ must hold. That is **three** equations for **two**
free mass parameters $(\mu_0^2,\mu_1^2)$ — the system is over-constrained, and this is the
point of the model rather than a defect. Solving two of them and substituting into the
third leaves a residual which must vanish on its own. `model.alignment_residual` computes
exactly that:


```python
tad = m.model.tadpoles()
print("the three tadpole conditions dV/dv_i = 0:")
for sym in (m.v1.s, m.v2.s, m.vS.s):
    print("  d/d%s:" % sym)
    display(sp.factor(sp.expand(tad[sym])))

residual = alignment_residual(m)
print("\nresidual of the third condition once the other two fix (mu0^2, mu1^2):")
display(sp.Eq(residual, 0))

roots = sp.solve(sp.Eq(residual, 0), m.v2.s)
print("\nroots for v2 (v1, v_S > 0):", roots)
assert sp.sqrt(3) * m.v1.s in roots
ok(r"the vacuum alignment v2 = sqrt(3) v1 is a *consequence* of the tadpole system")

ledger.step(r"The $\sqrt3$ vacuum alignment $v_2=\sqrt3\,v_1$ is forced by the tadpoles",
            obtained="derived",
            statement=r"\tfrac32\lambda_4v_S\,(3v_1^2-v_2^2)=0",
            checks=(r"derived from the Lagrangian's three tadpole conditions, not imported",
                    r"the corrected [DasDey14] erratum Eq. (9) has the same zero set (section 2.2)"),
            falsified_by=r"a tadpole residual vanishing off $v_2=\sqrt3v_1$ for $\lambda_4v_S\neq0$",
            section="2.1")
```

### 2.2 Cross-check: the erratum's corrected tadpoles give the same alignment

[DasDey14] carries an **Erratum, Phys. Rev. D 91, 039905 (2015)**, and its entire content is
a correction to Eqs. (9a)–(9c) — *these* tadpole conditions. It states: *"These changes,
however, do not affect any of our conclusions in the original paper."*

That is worth pausing on. The one equation Das & Dey had to correct is the one equation this
notebook never imported: §2.1 derives the tadpoles from the Lagrangian rather than
transcribing them. So the error could not reach us — but the *corrected* equations are now
an independent published object to check our derivation against.

Their Eqs. (9a) and (9b) are two expressions for the same $2\mu_1^2$, so the system is
over-constrained in exactly the way §2.1 describes, and their difference must vanish.

> **Before running the next cell.** Predict what setting (9a) $=$ (9b) gives. Our own
> residual was $\tfrac32\lambda_4v_S(3v_1^2-v_2^2)$, vanishing on $v_2=\sqrt3v_1$. Their
> paper labels the doublet VEVs the other way round (§2.1, §4), so what should their
> alignment read — and should the two residuals be *equal*, or only share a zero set?


```python
# ---- the erratum's corrected Eq. (9), transcribed verbatim ---------------
w1, w2, w3 = sp.symbols("v_1 v_2 v_3", positive=True)   # THEIR labels: v_3 is the singlet
# Use the MODEL's own lambda symbols, not freshly-made ones of the same name:
# two Symbols that print alike but carry different assumptions do NOT cancel,
# and the mismatch shows up only as an un-simplified ratio further down.
el = {k: m.lam_symbols[k - 1] for k in range(1, 9)}   # `lam` is not bound until §4
Bsum = el[5] + el[6] + 2*el[7]

eq9a = -2*el[1]*(w1**2 + w2**2) - 2*el[3]*(w1**2 + w2**2) - w3*(6*el[4]*w2 + Bsum*w3)
eq9b = (-2*el[1]*(w1**2 + w2**2) - 2*el[3]*(w1**2 + w2**2)
        - 3*w3/w2*el[4]*(w1**2 - w2**2) - Bsum*w3**2)
eq9c = el[4]*w2/w3*(w2**2 - 3*w1**2) - Bsum*(w1**2 + w2**2) - 2*el[8]*w3**2  # = 2 mu_3^2

# both (9a) and (9b) equal 2 mu_1^2, so the difference is their alignment residual
their_residual = sp.factor(sp.simplify(sp.expand(eq9a - eq9b)))
dv.show("[DasDey14] erratum: (9a) - (9b), which must vanish", their_residual,
        count_terms=False)

their_roots = [r for r in sp.solve(sp.Eq(their_residual, 0), w1) if r.is_positive]
print("their alignment:", their_roots, " -> v_1 = sqrt(3) v_2")
assert their_roots == [sp.sqrt(3) * w2]
ok("the erratum's corrected tadpoles force v_1 = sqrt(3) v_2")
```

```python
# ---- and that is OUR alignment, under the known v_1 <-> v_2 basis swap ---
ours = sp.factor(sp.expand(alignment_residual(m)))
swapped = sp.factor(sp.expand(ours.subs({m.v1.s: w2, m.v2.s: w1, m.vS.s: w3},
                                        simultaneous=True)))
dv.rows([(dv.lit(r"\text{ours, with }v_1\leftrightarrow v_2"), swapped),
         (dv.lit(r"\text{theirs, Eq.~(9a)}-\text{(9b)}"), their_residual)])

# the two are proportional, not equal: a tadpole residual is defined only up to
# normalisation (which combination of the three conditions you eliminate with).
ratio = sp.simplify(sp.cancel(their_residual / swapped))
print("theirs / ours =", ratio, "  (a nonzero factor, so the ZERO SETS coincide)")
assert ratio != 0 and ratio.free_symbols <= {w2}
ok("same zero set: our independently derived alignment and the erratum's corrected "
   "Eq. (9) agree, up to the documented v_1 <-> v_2 relabelling")

ledger.step(
    r"[DasDey14] Erratum, Phys. Rev. D 91, 039905 (2015) --- it corrects "
    r"Eqs. (9a)--(9c), the tadpole conditions, and nothing else",
    obtained="imported",
    checks=(r"the corrected (9a)$=$(9b) forces $v_1=\sqrt3v_2$",
            r"that is our own alignment residual under the documented "
            r"$v_1\leftrightarrow v_2$ swap, up to normalisation",
            r"Eq. (4) (boundedness) and Eqs. (36)--(37) (unitarity) are untouched, "
            r"so section 5 transcribes the final published text"),
    falsified_by=r"a corrected Eq. (9) whose zero set differs from our tadpole "
                 r"residual's",
    section="2.2")
```

The residual factorizes as $\tfrac32\lambda_4 v_S(3v_1^2-v_2^2)$, so — for $\lambda_4\neq0$
and $v_S\neq0$ — the vacuum must satisfy

$$v_2=\sqrt3\,v_1 .$$

This is [GomezBock21] Eq. (13). Their basis has the roles of $v_1$ and $v_2$ swapped
relative to feynlag's real-orthogonal irrep basis, so they write $v_1=\sqrt3\,v_2$; §4
shows why the two doublet bases differ.

### 2.3 The tutorial benchmark sits at the wrong electroweak scale

A second condition constrains the VEVs — the electroweak scale itself,

```python
v2_t, vS_t = 115.0, 80.0                 # the tutorial's numbers
v1_t = v2_t / math.sqrt(3)               # what `align` actually substitutes

print("before alignment: (200, 115, 80)        -> v = %6.2f GeV" %
      vacuum_scale(200.0, v2_t, vS_t))
print("after  alignment: (%.2f, %.1f, %.1f)   -> v = %6.2f GeV   <-- not 246" %
      (v1_t, v2_t, vS_t, vacuum_scale(v1_t, v2_t, vS_t)))
assert abs(vacuum_scale(v1_t, v2_t, vS_t) - V_EW) > 50
ok("confirmed: the tutorial's aligned vacuum misses the electroweak scale")
```

### 2.4 Imposing both conditions at once

The two conditions together leave a one-parameter family, which `aligned_vevs`
parametrizes by the same angle θ the geometric rotation uses ([GomezBock21] Eq. 24):

$$v_{12}\equiv\sqrt{v_1^2+v_2^2}=v\sin\theta,\qquad v_S=v\cos\theta,$$

and the alignment then fixes

$$v_1=\tfrac12 v_{12},\qquad v_2=\tfrac{\sqrt3}{2}v_{12}
\qquad\Longrightarrow\qquad \tan\varphi=\frac{v_2}{v_1}=\sqrt3,\quad\varphi=60°$$

for **every** θ — the doublet-plane angle is fixed by the alignment alone.


```python
for th in (0.15, 0.5, 1.0286, 1.3, 1.5):
    v1, v2, vS = aligned_vevs(th)
    assert abs(v2 / v1 - math.sqrt(3)) < 1e-12          # alignment
    assert abs(vacuum_scale(v1, v2, vS) - V_EW) < 1e-9  # electroweak scale
    print("theta = %.4f  ->  (%7.2f, %7.2f, %7.2f)   v = %.4f   v2/v1 = %.9f"
          % (th, v1, v2, vS, vacuum_scale(v1, v2, vS), v2 / v1))
ok("aligned_vevs satisfies BOTH vacuum conditions simultaneously, for every theta")

theta_t = math.atan2(math.hypot(v1_t, v2_t), vS_t)      # the tutorial's direction
v1, v2, vS = aligned_vevs(theta_t)
print("\nthe tutorial's direction, rescaled: (%.2f, %.2f, %.2f) at theta = %.2f deg"
      % (v1, v2, vS, math.degrees(theta_t)))
```

### 2.5 How much does this matter?

Every entry of every mass matrix is a quartic coupling times two VEVs, so each mass² is
**homogeneous of degree 2** in the VEVs. At fixed λ and θ the whole spectrum therefore just
rescales, by $(246/155.03)^2$. That is worth verifying rather than asserting:


```python
M_S, M_A, M_C = mass_matrices(m)
D_S, D_A, D_C = diagonal_blocks(m)

s = sp.Symbol("s", positive=True)
rescale = {m.v2.s: s * m.v2.s, m.vS.s: s * m.vS.s}
homogeneous = all(sp.simplify(D[i, j].subs(rescale) - s**2 * D[i, j]) == 0
                  for D in (D_S, D_A, D_C) for i in range(3) for j in range(3))
assert homogeneous
ok("every mass-matrix entry is homogeneous of degree 2 in the VEVs (27 entries)")

spec = spectrum_function(m)
naive_lams = [0.05 * k for k in range(1, 9)]        # the tutorial's benchmark lambdas
s_tut, s_fix = spec(naive_lams, v2_t, vS_t), spec(naive_lams, v2, vS)
expected = (V_EW / vacuum_scale(v1_t, v2_t, vS_t)) ** 2

print("\n%-6s %14s %14s %8s" % ("state", "tutorial m^2", "rescaled m^2", "ratio"))
for k in SPECTRUM_KEYS:
    print("%-6s %14.1f %14.1f %8.3f" % (k, s_tut[k], s_fix[k], s_fix[k] / s_tut[k]))
    assert abs(s_fix[k] / s_tut[k] - expected) < 1e-9
print("\nexpected ratio (246/155.03)^2 = %.3f" % expected)
ok("every mass is low by exactly 246/155.03 = %.3f in the tutorial" % math.sqrt(expected))
check_quoted("2.3, §9 and the report", [
    ("tutorial aligned vacuum [GeV]", vacuum_scale(v1_t, v2_t, vS_t), 155.03, 0.005),
    ("mass factor 246/155.03", math.sqrt(expected), 1.587, 0.0005)])
```

`model.py` lifts the tutorial's notebook-local algebra into importable functions. On the

```python
print("M_A, the CP-odd mass matrix on the aligned vacuum:")
display(sp.simplify(M_A))
assert M_A.is_symmetric() and M_C.is_symmetric() and M_S.is_symmetric()
ok("all three mass matrices are symmetric")
```

### 3.1 The geometric rotation

[GomezBock21] Eqs. (25)–(29) build a rotation out of the **vacuum geometry alone** — no
diagonalization involved:

$$\cos\varphi=\frac{v_1}{v_{12}},\quad \sin\varphi=\frac{v_2}{v_{12}},\qquad
\cos\theta=\frac{v_S}{v},\quad \sin\theta=\frac{v_{12}}{v},$$

$$R=\begin{pmatrix}
\sin\theta\cos\varphi & -\sin\varphi & -\cos\theta\cos\varphi\\
\sin\theta\sin\varphi & \phantom{-}\cos\varphi & -\cos\theta\sin\varphi\\
\cos\theta & 0 & \sin\theta
\end{pmatrix}.$$

Its first column *is* the vacuum direction, which is why $R^{\mathsf T}MR$ isolates the
Goldstone bosons at the $[0,0]$ entry.


```python
R = rotation(m)
print("R on the aligned vacuum (phi = 60 deg for every theta):")
display(R)

assert sp.simplify(R.T * R - sp.eye(3)) == sp.zeros(3, 3)
ok("R is exactly orthogonal:  R^T R = 1")
```

### 3.2 Goldstone counting

$SU(2)_L\times U(1)_Y\to U(1)_{\rm em}$ breaks **three** generators, so exactly three
Goldstone bosons must appear: one CP-odd ($G^0$) and one charged pair ($G^\pm$). They must
be **exactly** massless — not numerically small.


```python
def zero_pattern(D):
    return sp.Matrix(3, 3, lambda i, j: int(sp.simplify(D[i, j]) == 0))

for name, D in (("D_S", D_S), ("D_A", D_A), ("D_C", D_C)):
    print("%s  zero pattern (1 = vanishes):" % name)
    display(zero_pattern(D))

assert sp.simplify(D_A[0, 0]) == 0 and sp.simplify(D_C[0, 0]) == 0
for D in (D_A, D_C):                                   # fully diagonal
    assert all(sp.simplify(D[i, j]) == 0 for i in range(3) for j in range(3) if i != j)
ok("exactly one CP-odd + one charged Goldstone, both identically zero; D_A, D_C diagonal")

print("\nthe physical CP-odd and charged masses feynlag derives:")
for label, expr in ((r"m^2_{A_1}", D_A[1, 1]), (r"m^2_{A_2}", D_A[2, 2]),
                    (r"m^2_{H^\pm_1}", D_C[1, 1]), (r"m^2_{H^\pm_2}", D_C[2, 2])):
    display(sp.Eq(sp.Symbol(label), sp.simplify(expr)))
```

### 3.3 The leftover CP-even block

$R$ does **not** finish the CP-even sector: $D_S$ keeps one 2×2 block in the $(0,2)$
subspace, which the library's analytic 2×2 route closes with the standard

$$\tan 2\alpha=\frac{2\,B_{01}}{B_{00}-B_{11}} .$$


```python
block = cp_even_block(m)
print("the residual CP-even 2x2 block:")
display(sp.simplify(block))

theta_expr, tan2theta, _rot = cp_even_angle(m)
print("\nits mixing angle:")
display(sp.Eq(sp.Symbol(r"\tan 2\alpha"), sp.simplify(tan2theta)))

assert sp.simplify(D_S[0, 1]) == 0 and sp.simplify(D_S[1, 2]) == 0
assert sp.simplify(D_S[0, 2]) != 0
ok("D_S splits into h0 (already diagonal) plus one 2x2 block in the (0,2) subspace")
```

### 3.4 Independent check: brute-force diagonalization

Everything above rests on the *ansatz* that $R$ diagonalizes what feynlag derived. The

```python
results = check_sectors_against_eigenvals(m, naive_lams, v2, vS)
for sector, (via, brute, agree) in results.items():
    print("%-8s ansatz %s" % (sector, [round(x, 4) for x in via]))
    print("%-8s brute  %s   agree: %s" % ("", [round(x, 4) for x in brute], agree))
    assert agree
ok("the geometric ansatz reproduces direct numeric diagonalization in all three sectors")
```

## 4. Fixing the parameter basis against the literature

Published boundedness and unitarity conditions are stated in one specific parametrization;
applying them to a different one silently gives wrong answers. So the basis must be
*established*, not assumed. Three parametrizations of the same potential are in play:

| source | quartic parameters |
|---|---|
| `examples/thdm_s3.py` (feynlag) | $\lambda_1\ldots\lambda_8$ |
| [DasDey14] Eq. (3c) — where the constraints live | $\lambda_1\ldots\lambda_8$ |
| [GomezBock21] Eq. (2)/(6) | $a,b,c,d,e,f,g,h$ |

[GomezBock21] derives no bounds of its own; its §2.2 states plainly *"This analysis has
already been done in [66], and we use their expressions for the unitarity and stability
bounds in here"* — reference [66] being [DasDey14].

The tempting move is to declare feynlag's λ's identical to Das & Dey's because the written
invariants look alike. **That is wrong, and feynlag can prove it**: the λ₄ invariant
transcribed literally from [DasDey14] into feynlag's component labels is not S₃-invariant
in feynlag's real-orthogonal irrep basis.


```python
xS1, xS2 = bra(HS, H1), bra(HS, H2)
herm = lambda e: e + sp.conjugate(e)

variants = {
    "feynlag : xS1(x11-x22) - xS2(x12+x21)": herm(xS1 * (x11 - x22) - xS2 * (x12 + x21)),
    "DasDey  : xS1(x12+x21) + xS2(x11-x22)": herm(xS1 * (x12 + x21) + xS2 * (x11 - x22)),
}
verdict = {}
for label, expr in variants.items():
    verdict[label], _ = check_discrete_invariance(expr, m.s3)
    print("S3-invariant: %-5s   %s" % (verdict[label], label))

assert list(verdict.values()) == [True, False]
ok("the two doublet bases genuinely differ - one lambda4 structure is S3-invariant, the other is not")
```

### 4.1 What can differ between the two bases?

The bases are related by some rotation of the S₃ doublet $(H_1,H_2)$. Under a general
$SO(2)$ rotation by α, the singlet contraction $x_S=(x_{S1},x_{S2})$ rotates by **α** while
the doublet contraction $d_2$ rotates by **2α**. So most invariants are built from norms and
cannot change — but $x_S\cdot d_2$ mixes the two different rotation rates. Testing all eight
structures directly:


```python
al = sp.Symbol("alpha", real=True)
A1_, B1_, A2_, B2_, AS_, BS_ = sp.symbols("A1 B1 A2 B2 AS BS")
vbra = lambda X, Y: sp.conjugate(X[0]) * Y[0] + sp.conjugate(X[1]) * Y[1]

def structures(D1, D2, DS):
    y11, y22, y12, y21 = vbra(D1, D1), vbra(D2, D2), vbra(D1, D2), vbra(D2, D1)
    xS1, xS2, x1S, x2S = vbra(DS, D1), vbra(DS, D2), vbra(D1, DS), vbra(D2, DS)
    xSS = vbra(DS, DS)
    d0, d1 = y11 - y22, -(y12 + y21)
    return {
        "lambda1  (x11+x22)^2":   (y11 + y22) ** 2,
        "lambda2  (x12-x21)^2":   (y12 - y21) ** 2,
        "lambda3  |d2|^2":        d0**2 + d1**2,
        "lambda4  xS.d2 + h.c.":  herm(xS1 * d0 + xS2 * d1),
        "lambda5  xSS(x11+x22)":  xSS * (y11 + y22),
        "lambda6  xS1 x1S + xS2 x2S": xS1 * x1S + xS2 * x2S,
        "lambda7  xS1^2+xS2^2 + h.c.": herm(xS1**2 + xS2**2),
        "lambda8  xSS^2":         xSS ** 2,
    }

c_, s_ = sp.cos(al), sp.sin(al)
plain = structures((A1_, B1_), (A2_, B2_), (AS_, BS_))
rot = structures((c_ * A1_ - s_ * A2_, c_ * B1_ - s_ * B2_),
                 (s_ * A1_ + c_ * A2_, s_ * B1_ + c_ * B2_), (AS_, BS_))

invariant = {k: sp.simplify(sp.expand(rot[k] - plain[k])) == 0 for k in plain}
for k, v in invariant.items():
    print("  %-28s SO(2)-invariant: %s" % (k, v))

assert sum(invariant.values()) == 7 and not invariant["lambda4  xS.d2 + h.c."]
ok("7 of the 8 structures are basis-independent; ONLY the lambda4 structure can change")
```

That settles two things at once.

**(i) $a$, $b$, $c$ and the rest are fixed by term-matching.** Seven of the eight
structures are the *same object* in any doublet basis, so their coefficients transfer
directly. Reading them off the two potentials gives
$a=2\lambda_8$, $b=\lambda_5$, $c=2\lambda_1$, $d=2\lambda_2$, $f=\lambda_6$,
$g=2\lambda_3$, $h=2\lambda_7$.

**(ii) Only λ₄'s sign is at stake.** Because $x_S\cdot d_2\to\cos\alpha\,(x_S\cdot d_2)+
\sin\alpha\,(x_{S2}d_2^{(1)}-x_{S1}d_2^{(2)})$ and the second structure is *not*
S₃-invariant (that is exactly what the previous cell showed), the only α preserving S₃
invariance while changing anything is $\alpha=\pi$ — i.e. $H_1\to-H_1$, $H_2\to-H_2$ —
which flips $x_S\cdot d_2$ and leaves all seven others alone:


```python
flip = structures((-A1_, -B1_), (-A2_, -B2_), (AS_, BS_))
flipped = {k: sp.simplify(sp.expand(flip[k] - plain[k])) == 0 for k in plain}
for k, v in flipped.items():
    print("  %-28s unchanged by H1,H2 -> -H1,-H2: %s" % (k, v))

assert sum(flipped.values()) == 7 and not flipped["lambda4  xS.d2 + h.c."]
assert sp.simplify(sp.expand(flip["lambda4  xS.d2 + h.c."]
                             + plain["lambda4  xS.d2 + h.c."])) == 0
ok("H1,H2 -> -H1,-H2 flips exactly lambda4 and nothing else:  lambda4 -> -lambda4")
```

### 4.2 The sign: solve for the dictionary, do not guess it

§4.1 fixed seven of the eight entries by basis-independence and left exactly one open — the
sign of the λ₄ structure. [GomezBock21] publishes closed-form tree masses for the two
sectors that $R_A$ diagonalizes completely,

$$m^2_{A_1}=-v^2\Big[(d+g)\sin^2\theta+\tfrac54 e\sin2\theta+h\cos^2\theta\Big],
\qquad m^2_{A_2}=-v^2\Big[\tfrac{e}{2}\tan\theta+h\Big],$$

$$m^2_{H^\pm_1}=-\tfrac{v^2}{4}\Big[5e\sin2\theta+2(f+h)\cos^2\theta+4g\sin^2\theta\Big],
\qquad m^2_{H^\pm_2}=-\tfrac{v^2}{2}\Big[e\tan\theta+(f+h)\Big]$$

(Eqs. 30–33), and feynlag derived those same four masses independently in §3.2.

> **Before running the next cell.** The obvious move is to *try* $e=+\lambda_4$ and
> $e=-\lambda_4$ and keep whichever matches. Ask instead: with four published masses in the
> unknowns $(d,e,f,g,h)$, is the dictionary **over-determined**? If it is, you do not need to
> guess at all — you can solve, and the fact that a solution exists at all is itself a check
> on the whole chain. What would it mean if `solve` returned *no* solution, or a
> one-parameter family?

Match the four masses coefficient by coefficient in the independent trigonometric structures
and let sympy return the dictionary:


```python
v_sym, th_sym = sp.symbols("v theta", positive=True)
sub_vac = {m.v2.s: v_sym * sp.sin(th_sym) * sp.sqrt(3) / 2,
           m.vS.s: v_sym * sp.cos(th_sym)}

feynlag_masses = {
    "A1": sp.simplify(D_A[1, 1].subs(sub_vac)), "A2": sp.simplify(D_A[2, 2].subs(sub_vac)),
    "Hpm1": sp.simplify(D_C[1, 1].subs(sub_vac)), "Hpm2": sp.simplify(D_C[2, 2].subs(sub_vac)),
}
a_, b_, c2_, d_, e_, f_, g_, h_ = sp.symbols("a b c d e f g h")
gb21_masses = {
    "A1":   -v_sym**2 * ((d_ + g_) * sp.sin(th_sym)**2
                         + sp.Rational(5, 4) * e_ * sp.sin(2 * th_sym)
                         + h_ * sp.cos(th_sym)**2),                          # Eq. (30)
    "A2":   -v_sym**2 * (e_ / 2 * sp.tan(th_sym) + h_),                      # Eq. (31)
    "Hpm1": -v_sym**2 / 4 * (5 * e_ * sp.sin(2 * th_sym)
                             + 2 * (f_ + h_) * sp.cos(th_sym)**2
                             + 4 * g_ * sp.sin(th_sym)**2),                  # Eq. (32)
    "Hpm2": -v_sym**2 / 2 * (e_ * sp.tan(th_sym) + (f_ + h_)),               # Eq. (33)
}
lam = m.lam_symbols

# ---- MOVE 1: set up.  Look at one published mass and feynlag's beside it. --
print("A_2, as published [GomezBock21] Eq. (31)  and  as feynlag derives it:")
display(sp.Eq(sp.Symbol("m^2_{A_2}"), gb21_masses["A2"]))
display(sp.Eq(sp.Symbol("m^2_{A_2}"), sp.simplify(feynlag_masses["A2"])))
```

The two are the same function of θ, so requiring them equal is four *identities in θ*, not
four numbers — and an identity has to hold at **every** θ. Evaluating each of the four at
several angles therefore manufactures as many linear equations in $(d,e,f,g,h)$ as we care
to write down. Five unknowns, twenty equations: badly over-determined, so `solve` either
returns one dictionary or none at all.


```python
# ---- MOVE 2: collect.  An identity in theta holds at every theta, so ------
# evaluating at several angles turns it into a linear system for (d,e,f,g,h).
unknowns = [d_, e_, f_, g_, h_]
angles = [sp.pi/6, sp.pi/4, sp.pi/3, sp.Rational(2, 5), sp.Rational(7, 10)]
eqs = [sp.Eq(sp.simplify((feynlag_masses[k] - gb21_masses[k]).subs(
                 {th_sym: a, v_sym: 1})), 0)
       for k in feynlag_masses for a in angles]
print("%d equations for %d unknowns -- over-determined by a wide margin"
      % (len(eqs), len(unknowns)))

solution = sp.solve(eqs, unknowns, dict=True)
assert len(solution) == 1, f"expected a unique dictionary, got {len(solution)}"
sol = solution[0]
dv.show("the dictionary sympy returns", sp.Matrix(
    [[u, sp.simplify(sol[u])] for u in unknowns]), count_terms=False)
```

The sign is an **output**. Nothing was tried and kept: with the seven basis-independent
entries of §4.1 already fixed, the published masses determine the eighth, and

$$\boxed{e=-\lambda_4}$$

falls out. That the over-determined system has *any* solution is the real check here — it
says feynlag's potential, tadpoles, mass matrices and rotation all agree with four
independently published closed forms simultaneously. A single error anywhere in that chain
would have left the system inconsistent and `solve` would have returned nothing.

Now the checks:


```python
# ---- MOVE 3 + 4: recognise, then check ------------------------------------
assert sp.simplify(sol[e_] + lam[3]) == 0
ok("solved, not guessed:  e = -lambda_4")
for sym, expect, name in ((d_, 2*lam[1], "d = 2 lambda_2"), (f_, lam[5], "f = lambda_6"),
                          (g_, 2*lam[2], "g = 2 lambda_3"), (h_, 2*lam[6], "h = 2 lambda_7")):
    assert sp.simplify(sol[sym] - expect) == 0, f"{name} disagrees: {sol[sym]}"
    print("   %s" % name)
ok("the other four solved entries agree with the term-matching of section 4.1")

# check: the WRONG sign must leave the system inconsistent, not merely unequal
wrong = {a_: 2*lam[7], b_: lam[4], c2_: 2*lam[0], d_: 2*lam[1],
         e_: +lam[3], f_: lam[5], g_: 2*lam[2], h_: 2*lam[6]}
bad = [sp.simplify(sp.expand_trig(sp.simplify(
    feynlag_masses[k] - gb21_masses[k].subs(wrong)))) for k in feynlag_masses]
assert all(b != 0 for b in bad)
ok("e = +lambda_4 fails ALL FOUR masses, not just one -- the sign is not marginal")

# check: switch lambda_4 off and the published form must still be reproduced --
# a limit where the sign under debate cannot matter, so it tests the REST of
# the dictionary independently of the thing we just solved for.
dv.check_limit(sp.simplify(feynlag_masses["A2"].subs(lam[3], 0)), {},
               sp.simplify(gb21_masses["A2"].subs({e_: 0, h_: 2 * lam[6]})),
               "at lambda_4 = 0 the published m^2_{A_2} is still reproduced")

ledger.step(r"The [GomezBock21] $(a\ldots h)\leftrightarrow$ feynlag $\lambda$ "
            r"dictionary, with $e=-\lambda_4$",
            obtained="derived",
            expr=sp.Matrix([[u, sol[u]] for u in unknowns]),
            checks=(r"solved from an over-determined linear system, not guessed",
                    r"the four solved entries agree with §4.1's term matching",
                    r"$e=+\lambda_4$ fails all four published masses",
                    r"the sign question vanishes in the $\lambda_4\to0$ limit"),
            falsified_by=r"$\mathrm{solve}()$ returning no solution, which any error "
                         r"in the potential/tadpole/mass-matrix/rotation chain "
                         r"would cause",
            section="4.2")
```

So the dictionary is

$$a=2\lambda_8,\quad b=\lambda_5,\quad c=2\lambda_1,\quad d=2\lambda_2,\quad
\boxed{e=-\lambda_4},\quad f=\lambda_6,\quad g=2\lambda_3,\quad h=2\lambda_7 ,$$

seven entries fixed by the basis-independence argument of §4.1 and the last by matching
published masses. Composing with the [GomezBock21]↔[DasDey14] correspondence read off the
two potentials ($\lambda^{\rm DD}_1=c/2$, $\lambda^{\rm DD}_2=d/2$, $\lambda^{\rm DD}_3=g/2$,
$\lambda^{\rm DD}_4=e$, $\lambda^{\rm DD}_5=b$, $\lambda^{\rm DD}_6=f$,
$\lambda^{\rm DD}_7=h/2$, $\lambda^{\rm DD}_8=a/2$):

$$\lambda^{\rm DasDey}_k=\lambda^{\rm feynlag}_k\ \ \text{for all } k\neq4,
\qquad \lambda^{\rm DasDey}_4=-\lambda^{\rm feynlag}_4 .$$

**And λ₄ enters every constraint below only as $|\lambda_4|$ or $\lambda_4^2$** — which is
no accident, but the direct consequence of §4.1: the field redefinition $H_{1,2}\to-H_{1,2}$
is a symmetry of the theory that flips λ₄, so no physical condition may depend on its sign.
The conditions therefore transfer with no correction at all.

As a by-product, the agreement of all four masses validates feynlag's entire 3HDM-S₃ chain
— potential, tadpoles, mass matrices and rotation — against published closed forms.

## 5. Theory constraints

Two requirements, both quoted from [DasDey14] and implemented in `constraints.py`.

> ✅ **Provenance, settled.** [DasDey14] carries an **Erratum, Phys. Rev. D 91, 039905
> (2015)**, never posted to arXiv (the arXiv record stops at v2, May 2014). It has now been
> read: its entire content corrects **Eqs. (9a)–(9c)**, the tadpole conditions checked in
> §2.2, and it states that *"these changes do not affect any of our conclusions"*.
> **Eq. (4) and Eqs. (36)–(37) below are untouched by it**, so what follows is the final
> published form, not a superseded one. §5.3 nevertheless does not take the conditions on
> trust — it tests them directly against the potential, and finds a real problem.

**Boundedness from below**, [DasDey14] Eq. (4a)–(4g):

$$\lambda_1>0,\qquad \lambda_8>0,\qquad \lambda_1+\lambda_3>0,\qquad
2\lambda_1+(\lambda_3-\lambda_2)>|\lambda_2+\lambda_3|,$$
$$\lambda_5+2\sqrt{\lambda_8(\lambda_1+\lambda_3)}>0,\qquad
\lambda_5+\lambda_6+2\sqrt{\lambda_8(\lambda_1+\lambda_3)}>2|\lambda_7|,$$
$$\lambda_1+\lambda_3+\lambda_5+\lambda_6+2\lambda_7+\lambda_8>2|\lambda_4| .$$

**Tree-level unitarity**, [DasDey14] Eq. (36) with the eigenvalues (37a)–(37l): every
eigenvalue of the $\ell=0$ two-body S-matrix must satisfy $|a|\le16\pi$. Twelve of them come
in pairs $a^\pm=X\pm\sqrt{X^2-4Y}$,

$$
\begin{aligned}
&X_{1}=\lambda_1-\lambda_2+\tfrac{\lambda_5+\lambda_6}{2}, &\ &Y_{1}=(\lambda_1-\lambda_2)\tfrac{\lambda_5+\lambda_6}{2}-\lambda_4^2,\\
&X_{2}=\lambda_1+\lambda_2+2\lambda_3+\lambda_8, &\ &Y_{2}=\lambda_8(\lambda_1+\lambda_2+2\lambda_3)-2\lambda_7^2,\\
&X_{3}=\lambda_1-\lambda_2+2\lambda_3+\lambda_8, &\ &Y_{3}=\lambda_8(\lambda_1-\lambda_2+2\lambda_3)-\tfrac{\lambda_4^2}{2},\\
&X_{4}=\lambda_1+\lambda_2+\tfrac{\lambda_5}{2}+\lambda_7, &\ &Y_{4}=(\lambda_1+\lambda_2)(\tfrac{\lambda_5}{2}+\lambda_7)-\lambda_4^2,\\
&X_{5}=5\lambda_1-\lambda_2+2\lambda_3+3\lambda_8, &\ &Y_{5}=3\lambda_8(5\lambda_1-\lambda_2+2\lambda_3)-\tfrac{(2\lambda_5+\lambda_6)^2}{2},\\
&X_{6}=\lambda_1+\lambda_2+4\lambda_3+\tfrac{\lambda_5}{2}+\lambda_6+3\lambda_7, &\ &Y_{6}=(\lambda_1+\lambda_2+4\lambda_3)(\tfrac{\lambda_5}{2}+\lambda_6+3\lambda_7)-9\lambda_4^2,
\end{aligned}
$$

and six are already diagonal:

$$b_1=\lambda_5+2\lambda_6-6\lambda_7,\quad b_2=\lambda_5-2\lambda_7,\quad
b_3=2(\lambda_1-5\lambda_2-2\lambda_3),$$
$$b_4=2(\lambda_1-\lambda_2-2\lambda_3),\quad b_5=2(\lambda_1+\lambda_2-2\lambda_3),\quad
b_6=\lambda_5-\lambda_6 .$$


### 5.1 The typeset equations and the executed code must agree

Displayed maths and running code drift apart. Rebuilding the eigenvalues above
independently in SymPy and comparing against `constraints.unitarity_eigenvalues` closes
that gap. The same cell checks the claim that every discriminant $X^2-4Y$ is a **sum of
squares**, hence never negative — so the square roots need no branch guard.


```python
l1, l2, l3, l4, l5, l6, l7, l8 = sp.symbols("lambda1:9", real=True)
XY = [
    (l1 - l2 + (l5 + l6) / 2,               (l1 - l2) * (l5 + l6) / 2 - l4**2),
    (l1 + l2 + 2*l3 + l8,                   l8 * (l1 + l2 + 2*l3) - 2 * l7**2),
    (l1 - l2 + 2*l3 + l8,                   l8 * (l1 - l2 + 2*l3) - l4**2 / 2),
    (l1 + l2 + l5/2 + l7,                   (l1 + l2) * (l5/2 + l7) - l4**2),
    (5*l1 - l2 + 2*l3 + 3*l8,               3*l8 * (5*l1 - l2 + 2*l3) - (2*l5 + l6)**2 / 2),
    (l1 + l2 + 4*l3 + l5/2 + l6 + 3*l7,     (l1 + l2 + 4*l3) * (l5/2 + l6 + 3*l7) - 9*l4**2),
]
sos = [(l1 - l2 - (l5 + l6)/2)**2 + 4*l4**2,
       (l1 + l2 + 2*l3 - l8)**2 + 8*l7**2,
       (l1 - l2 + 2*l3 - l8)**2 + 2*l4**2,
       (l1 + l2 - l5/2 - l7)**2 + 4*l4**2,
       (5*l1 - l2 + 2*l3 - 3*l8)**2 + 2*(2*l5 + l6)**2,
       (l1 + l2 + 4*l3 - l5/2 - l6 - 3*l7)**2 + 36*l4**2]

for i, ((X, Y), S) in enumerate(zip(XY, sos)):
    assert sp.expand(X**2 - 4*Y - S) == 0
    print("  Eq.(37%s) discriminant = %s" % ("abcdef"[i], S))
ok("every discriminant is a manifest sum of squares - always real, no branch guard needed")

typeset = []
for X, Y in XY:
    typeset += [X + sp.sqrt(X**2 - 4*Y), X - sp.sqrt(X**2 - 4*Y)]
typeset += [l5 + 2*l6 - 6*l7, l5 - 2*l7, 2*(l1 - 5*l2 - 2*l3),
            2*(l1 - l2 - 2*l3), 2*(l1 + l2 - 2*l3), l5 - l6]
fn = sp.lambdify([l1, l2, l3, l4, l5, l6, l7, l8], typeset, "numpy")

rng = np.random.default_rng(3)
pts = rng.uniform(-3, 3, size=(200, 8))
delta = max(np.abs(np.array(fn(*p), float) - unitarity_eigenvalues(p[None, :])[0]).max()
            for p in pts)
print("\nmax |typeset - constraints.py| over 200 random points: %.2e" % delta)
assert delta < 1e-12
ok("the 18 typeset eigenvalues are exactly what constraints.py computes")
```

### 5.3 Are the boundedness conditions actually sufficient?

Boundedness from below is a statement about the quartic part of the potential: $V_4$ is
homogeneous of degree 4 in the fields, so **the potential is bounded from below iff
$V_4\ge0$ everywhere on the unit sphere in field space**. That is directly testable, with
no reference to any paper. Splitting each of the six complex components as $x_i+iy_i$ gives
12 real degrees of freedom.

> **Work this one out before reading on.** [DasDey14] Eq. (4g) is a *single* inequality,
> $\lambda_1+\lambda_3+\lambda_5+\lambda_6+2\lambda_7+\lambda_8>2|\lambda_4|$. In a moment
> we will restrict $V_4$ to real neutral fields, where it becomes a quartic in essentially
> two variables. Ask yourself: **how many independent conditions does it take for such a
> quartic to be non-negative *everywhere*, and what would have to be true of the potential
> for one inequality to suffice?** Commit to an answer — the next three cells settle it,
> and if your answer was "one" the reason it fails is the interesting part.

First, confirm the extracted $V_4$ is the right object:


```python
V = -sum(t.expr for t in m.model.lagrangian.terms)
V4 = sum(t for t in sp.expand(V).args if any(t.has(l) for l in lam))
xr, yi = sp.symbols("x0:6", real=True), sp.symbols("y0:6", real=True)
split = {c: xr[i] + sp.I * yi[i] for i, c in enumerate(component_symbols(m))}
re_part, im_part = sp.expand(V4.subs(split)).as_real_imag()
re_part = sp.expand(re_part)

assert sp.simplify(im_part) == 0
ok("V4 is real once the components are split into x + i y")

# the scaling reflex: degree-4 homogeneity is what licenses "bounded from below
# iff non-negative on the unit sphere" -- the radial direction factors out.
dv.check_homogeneous(re_part, list(xr) + list(yi), 4,
                     "V4 is homogeneous of degree 4 in the 12 real components")

zero = {v: 0 for v in list(xr) + list(yi)}
spot = {"only H_S real  -> lambda8":         ({xr[5]: 1}, lam[7]),
        "only H_1 real  -> lambda1+lambda3": ({xr[1]: 1}, lam[0] + lam[2]),
        "only H_2 real  -> lambda1+lambda3": ({xr[3]: 1}, lam[0] + lam[2])}
for label, (pt, expect) in spot.items():
    assert sp.simplify(sp.expand(re_part.subs({**zero, **pt})) - expect) == 0
    print("  %-36s %s" % (label, sp.simplify(expect)))
ok("V4 reproduces the hand-derived value along three independent single-field directions")
```

Now sample random unit directions at λ points that Eq. (4) **accepts**. Any negative value
found is a proof of unboundedness — an explicit direction, and $V_4$ being degree-4
homogeneous the potential runs to $-\infty$ along it.


```python
rng = np.random.default_rng(11)
pool = rng.uniform(-2.0, 2.0, size=(300_000, 8))
accepted = pool[bfb_mask(pool)]
print("lambda points satisfying [DasDey14] Eq.(4):", len(accepted))

probe = accepted[:400]
worst = numeric_bfb_min(m, probe, n_directions=40_000, seed=1, subspace="neutral-real")
nbad = int((worst < -1e-9).sum())
print("of %d such points, %d have a REAL NEUTRAL direction with V4 < 0  (%.1f%%)"
      % (len(probe), nbad, 100 * nbad / len(probe)))
print("most negative V4 found: %.4f" % worst.min())
assert nbad > 0, "expected the quoted conditions to admit unbounded points"
ok("the quoted Eq.(4) conditions are NOT sufficient - counterexamples exist")
```

The failure is not some exotic charged direction: it already shows up in the **real neutral
slice**, the smallest subspace any boundedness analysis covers. Here is one point in full.


```python
j = int(np.argmin(worst))
lam_bad = probe[j]
r1s, r2s, rSs = sp.symbols("r_1 r_2 r_S", real=True)
V4n = sp.expand(re_part.subs({**zero, xr[1]: r1s, xr[3]: r2s, xr[5]: rSs}))
fn_n = sp.lambdify([r1s, r2s, rSs] + lam, V4n, "numpy")

ang = np.random.default_rng(2).uniform(0, 2*np.pi, size=(2, 200_000))
d1, d2v, dS = (np.sin(ang[0])*np.cos(ang[1]), np.sin(ang[0])*np.sin(ang[1]), np.cos(ang[0]))
vals = fn_n(d1, d2v, dS, *lam_bad)
k = int(np.argmin(vals))
direction = (float(d1[k]), float(d2v[k]), float(dS[k]))

print("lambda      =", np.round(lam_bad, 4).tolist())
print("Eq.(4a-4g)  =", bfb_conditions(lam_bad[None, :])[0].astype(int),
      "-> all satisfied:", bool(bfb_mask(lam_bad[None, :])[0]))
print("direction (r1, r2, rS) =", tuple(round(x, 4) for x in direction),
      "  |r| = %.6f" % float(np.linalg.norm(direction)))
print("V4 along it = %.6f      (so V4 -> -infinity as t^4)" % vals[k])
assert bool(bfb_mask(lam_bad[None, :])[0]) and vals[k] < 0
ok("explicit counterexample: Eq.(4) satisfied, yet the potential is unbounded below")
```

### 5.4 Diagnosis: derive the reduction, do not assume it

Now the actual piece of work. Restrict $V_4$ to real neutral components
$(H_1^0,H_2^0,H_S^0)=(r_1,r_2,r_S)$ and **look at what you get** before naming anything.
This is the step you would do on paper: write the restricted potential out, then collect.


```python
# ---- MOVE 1: set up.  Write it out and look at it. -----------------------
dv.show("V4 restricted to the real neutral slice", V4n, collect=rSs)
```

Fifteen terms, and no structure is obvious yet. The move on paper is to ask **what
multiplies each coupling** — because two couplings carrying the *same* coefficient can only
ever appear in that one combination, and that is what collapses the problem.


```python
# ---- MOVE 2: collect.  Let the structure emerge; do not type in the answer.
coeffs = dv.coefficients_of(V4n, lam)

print("\nreading it off:")
groups = {}
for p, c in coeffs.items():
    groups.setdefault(sp.srepr(sp.expand(c)), []).append(p)
for _, g in groups.items():
    if len(g) > 1:
        print("   identical coefficient ->", g, " so only their SUM can appear")
print("   lambda_2 multiplies %s  <- it drops out of this slice entirely"
      % coeffs[lam[1]])
```

There is the whole reduction, read off rather than asserted:

- $\lambda_1$ and $\lambda_3$ carry the **same** coefficient $(r_1^2+r_2^2)^2$, so only
  $\lambda_1+\lambda_3$ can ever appear;
- $\lambda_5,\lambda_6$ carry $r_S^2(r_1^2+r_2^2)$ and $\lambda_7$ carries twice it, so only
  $\lambda_5+\lambda_6+2\lambda_7$ appears;
- $\lambda_8$ carries $r_S^4$;
- **$\lambda_2$ multiplies zero.** It is the coefficient of $(x_{12}-x_{21})^2$, which needs
  a *relative phase* between the two doublets — a real neutral slice simply cannot see it.
  That is why $\lambda_2$ never shows up in $f(t)$ below, and it is worth knowing rather
  than wondering about later.

So with $x\equiv r_1^2+r_2^2$ and $y\equiv r_S^2$ everything is a function of $(x,y)$ —
*except* the $\lambda_4$ term, which is $2\lambda_4\,r_S\,r_1(r_1^2-3r_2^2)$ and is not a
function of $x$ alone. That is the term that will break the single-inequality argument, so
it deserves a name.


```python
# ---- MOVE 3: recognise.  What IS that lambda_4 structure? ----------------
covariant = sp.factor(sp.expand(coeffs[lam[3]] / (2 * rSs)))
print("lambda_4 coefficient / (2 r_S) =", covariant)
print("Re[(r_1 + i r_2)^3]            =", sp.expand(sp.re(sp.expand((r1s + sp.I*r2s)**3))))
assert sp.simplify(covariant - sp.re(sp.expand((r1s + sp.I*r2s)**3))) == 0
ok("it is Re[(r_1 + i r_2)^3] -- the Z_3 cubic covariant, not an accident")
```

$r_1(r_1^2-3r_2^2)=\operatorname{Re}\big[(r_1+ir_2)^3\big]$ is the **cubic covariant of the
$\mathbb Z_3\subset S_3$** acting on the doublet. Writing $r_1=\sqrt x\cos\psi$,
$r_2=\sqrt x\sin\psi$ it is $x^{3/2}\cos3\psi$ — so as $\psi$ ranges over the circle this
term sweeps $\pm x^{3/2}$, and its worst case is $-2|\lambda_4|\,x^{3/2}\sqrt y$.

**That is the crux.** The $(x,y)$ part is a quadratic form, but the $\lambda_4$ term carries
a half-integer power of $x$, so the whole thing is a *quartic in $t=\sqrt{y/x}$*, not a
quadratic form — and a quartic needs more than one inequality.

Now the checks, none of which re-runs the algebra above:


```python
# ---- MOVE 4: check, from physics rather than from the same algebra -------
x, y = sp.symbols("x y", nonnegative=True)

# (a) scaling: the slice inherits degree-4 homogeneity
dv.check_homogeneous(V4n, [r1s, r2s, rSs], 4,
                     "the restricted V4 is still homogeneous of degree 4")

# (b) symmetry: the covariant must be Z_3-invariant (rotation by 2*pi/3)
c3, s3 = sp.Rational(-1, 2), sp.sqrt(3) / 2
z3 = {r1s: c3*r1s - s3*r2s, r2s: s3*r1s + c3*r2s}
dv.check_invariant(covariant, z3, "the cubic covariant is invariant under Z_3 (2pi/3)")

# (c) the limit that explains why Eq.(4g) LOOKED right: switch lambda_4 off
A_, B_ = lam[0] + lam[2], lam[4] + lam[5] + 2 * lam[6]
dv.check_limit(V4n.subs(lam[3], 0), {r1s: sp.sqrt(x), r2s: 0, rSs: sp.sqrt(y)},
               A_*x**2 + B_*x*y + lam[7]*y**2,
               "lambda_4 -> 0 leaves a pure quadratic form in (x, y)")

# (d) a slice small enough to finish on paper
_ = dv.hand_slice(V4n, {r2s: 0}, "hand slice: set r_2 = 0", collect=rSs)
```

Check (c) is the one that earns its keep. With $\lambda_4=0$ the potential is a genuine
**quadratic form** in $(x,y)$ with $x,y\ge0$, and non-negativity of such a form is the
textbook copositivity condition — three inequalities, of which the cross-term one is
*exactly* Eq. (4g). So the published condition is not careless: it is correct in the
$\lambda_4\to0$ limit, and the half-integer power that $\lambda_4$ introduces is precisely
what invalidates it. That is worth more than knowing the answer.

Check (d) leaves $V_4\big|_{r_2=0}=(\lambda_1+\lambda_3)r_1^4+2\lambda_4r_1^3r_S+
(\lambda_5+\lambda_6+2\lambda_7)r_1^2r_S^2+\lambda_8r_S^4$ — four terms, two variables,
minimisable by hand in a couple of minutes. Doing that on paper is the fastest way to
convince yourself the cubic cross-term really can drive it negative.

Finally, confirm the grouped form matches the raw expression — this is now a *confirmation*
of something derived, not the derivation itself:


```python
grouped = (A_ * (r1s**2 + r2s**2)**2 + B_ * (r1s**2 + r2s**2) * rSs**2
           + lam[7] * rSs**4 + 2 * lam[3] * rSs * covariant)
assert sp.simplify(sp.expand(V4n - grouped)) == 0
ok("the grouped form reproduces the raw restricted V4 exactly")

# and the polar form that gives the worst case over psi
Xp, psi = sp.Symbol("X", positive=True), sp.Symbol("psi", real=True)
polar = sp.simplify(sp.expand_trig(sp.expand(
    covariant.subs({r1s: sp.sqrt(Xp)*sp.cos(psi), r2s: sp.sqrt(Xp)*sp.sin(psi)}))))
assert sp.simplify(polar - Xp**sp.Rational(3, 2) * sp.cos(3*psi)) == 0
ok("in polar form it is x^(3/2) cos(3 psi), so the worst case over psi is -2|lambda4|")

# a check that cannot fail is not a check: break the covariant's relative sign
dv.falsify("Z_3 invariance of the cubic covariant",
           lambda: dv.check_invariant(r1s*(r1s**2 + 3*r2s**2), z3, "(broken)"))

ledger.step(
    r"The real neutral slice, and the absence of $\lambda_2$",
    obtained="derived",
    statement=r"V_4 = (\lambda_1+\lambda_3)\,x^2 + (\lambda_5+\lambda_6+2\lambda_7)\,xy"
              r" + \lambda_8\,y^2 + 2\lambda_4\,\sqrt{y}\,x^{3/2}\cos3\psi",
    expr=grouped,
    checks=(r"degree-4 homogeneous",
            r"the cubic covariant is $\mathbb{Z}_3$-invariant "
            r"(falsified on a broken sign)",
            r"$\lambda_4\to0$ reduces to a quadratic form in $(x,y)$",
            r"the $r_2=0$ hand slice reproduces it"),
    falsified_by=r"any term outside the $(x,y,\cos3\psi)$ basis, or $\lambda_2$ "
                 r"appearing",
    section="5.4")
```

Every term is degree 2 in $(x,y)$, so setting $x=1$ and $t=\sqrt y$ loses nothing. The
**exact** boundedness condition on this slice is therefore

$$\boxed{\;f(t)\;=\;\lambda_8 t^4+(\lambda_5+\lambda_6+2\lambda_7)\,t^2-2|\lambda_4|\,t
+(\lambda_1+\lambda_3)\;\ge\;0\qquad\text{for all }t\ge0\;}$$

and now the problem is plain. **[DasDey14] Eq. (4g),**

$$\lambda_1+\lambda_3+\lambda_5+\lambda_6+2\lambda_7+\lambda_8>2|\lambda_4|,$$

**is exactly $f(1)>0$** — a single point of a curve that has to stay non-negative
everywhere. It is necessary, not sufficient.


```python
A_n = float(lam_bad[0] + lam_bad[2])
B_n = float(lam_bad[4] + lam_bad[5] + 2 * lam_bad[6])
C_n, l8_n = 2 * abs(float(lam_bad[3])), float(lam_bad[7])
f = lambda t: l8_n * t**4 + B_n * t**2 - C_n * t + A_n

ts = np.linspace(0, 1.6, 400)
fig, ax = plt.subplots(figsize=(6.0, 3.2))
ax.plot(ts, f(ts), color="#4C72B0", lw=1.8, label=r"$f(t)$")
ax.axhline(0, color="black", lw=0.8)
ax.fill_between(ts, f(ts), 0, where=f(ts) < 0, color="#C44E52", alpha=0.3,
                label=r"$f(t)<0$: unbounded")
ax.plot([1], [f(1)], "o", color="#55A868", ms=7, zorder=5,
        label=r"Eq.(4g) tests only $t=1$:  $f(1)=%.3f>0$" % f(1))
ax.set_xlabel(r"$t=\sqrt{y}=|r_S|$  (at $x=r_1^2+r_2^2=1$)")
ax.set_ylabel(r"$f(t)$")
ax.set_title("Eq. (4g) checks one point of a curve")
ax.legend(fontsize=8)
plt.tight_layout(); plt.show()

print("min over t >= 0 : %.4f  at t = %.3f" % (f(ts).min(), ts[int(np.argmin(f(ts)))]))
print("closed form     : %.4f" % neutral_real_bfb_min(lam_bad[None, :])[0])
assert f(1) > 0 > f(ts).min()
ok("Eq.(4g) passes at t=1 while f dips below zero elsewhere - exactly the counterexample")
```

### 5.5 The corrected condition

`constraints.neutral_real_bfb_min` minimizes $f(t)$ over $t\ge0$ in closed form (the
stationary points are the real positive roots of the cubic $f'$, found with batched
companion matrices), and `strict_bfb_mask` = Eq. (4) **and** $\min_t f\ge0$.

Its validation is the brute-force direction scan on the same slice — the two must agree on
every point, since one solves exactly what the other samples:


```python
closed = neutral_real_bfb_min(probe)
brute = numeric_bfb_min(m, probe, n_directions=60_000, seed=4, subspace="neutral-real")
disagree = int(((brute < -1e-6) != (closed < -1e-9)).sum())
print("closed form vs 60k-direction brute force over %d points: %d disagreements"
      % (len(probe), disagree))
assert disagree == 0
ok("the closed-form neutral condition reproduces brute-force sampling exactly")

print("\nimpact on the quoted conditions:")
print("  Eq.(4) acceptance                     : %.3f%%" % (100 * bfb_mask(pool).mean()))
print("  Eq.(4) + corrected neutral condition  : %.3f%%" % (100 * strict_bfb_mask(pool).mean()))
print("  fraction of Eq.(4) survivors removed  : %.2f%%"
      % (100 * (1 - strict_bfb_mask(pool).sum() / bfb_mask(pool).sum())))
check_quoted("5.5, §9 and the report", [
    ("Eq.(4) survivors in the 3e5-point pool", int(bfb_mask(pool).sum()), 17277, 0),
    ("Eq.(4) survivors removed by the corrected condition [%]",
     100 * (1 - strict_bfb_mask(pool).sum() / bfb_mask(pool).sum()), 9.35, 0.005)])
```

**Scope, stated honestly.** The corrected condition is exact on the real neutral slice, so
it is strictly stronger than Eq. (4g) — but it is still only *necessary*, not sufficient:
complex-neutral and charged directions are not covered by it. Full necessary-and-sufficient
BFB conditions for a 3HDM are a research problem in their own right (compare
[BotoRomaoSilva22], which solves other symmetry classes). So the scan below also carries an
**empirical backstop**: `model.numeric_bfb_min` samples the full 12-dimensional field space
on the surviving points, and any point with a negative direction is discarded.

It is now settled that the 2015 erratum does **not** correct Eq. (4g): it corrects only
the tadpole equations (§2.2). So the insufficiency demonstrated here is an insufficiency of
the conditions as they stand in the final published record — not of a superseded
pre-erratum form, and not a rediscovery of a known correction.

## 6. The scan

Nine parameters: the eight quartics and the vacuum angle θ.

- $\lambda_i\in U(-2,2)$ — deliberately **not** the tutorial's $U(0,1.5)$. Boundedness only
  requires $\lambda_1,\lambda_8>0$; forcing all eight positive presupposes part of the
  answer. The range stays well inside perturbativity ($|\lambda|<4\pi$).
- $\theta\in U(0.05,\pi/2-0.05)$, avoiding the endpoints where a VEV vanishes.

Every point is evaluated with numpy arrays — no Python loop over the sample.


```python
rng = np.random.default_rng(20260729)
N = 2_000_000

lam_scan = rng.uniform(-2.0, 2.0, size=(N, 8))
theta = rng.uniform(0.05, np.pi / 2 - 0.05, size=N)
v12_s = V_EW * np.sin(theta)
v2_s, vS_s = v12_s * np.sqrt(3) / 2, V_EW * np.cos(theta)

vspec = spectrum_function(m, "numpy")
t0 = time.time()
msq = vspec([lam_scan[:, i] for i in range(8)], v2_s, vS_s)
print("spectrum for %d points in %.2f s" % (N, time.time() - t0))
```

The scan rests entirely on the vectorized spectrum, so it must reproduce the scalar one.
Note what this does and does not establish: both sides lambdify the *same* expression, so it
checks the broadcasting plumbing, not the physics. The independent physics check is the
brute-force diagonalization of §3.4, repeated on the quoted benchmarks in §8.


```python
scal = spectrum_function(m, "math")
idx_chk = rng.integers(0, N, size=25)
worst_rel = 0.0
for i in idx_chk:
    a = scal(list(lam_scan[i]), float(v2_s[i]), float(vS_s[i]))
    for key in SPECTRUM_KEYS:
        b = float(msq[key][i])
        worst_rel = max(worst_rel, abs(a[key] - b) / max(abs(a[key]), 1.0))
print("max relative difference (numpy vs math) over 25 random points: %.2e" % worst_rel)
assert worst_rel < 1e-10
ok("the vectorized spectrum agrees with the scalar one")
```

```python
no_tachyon = np.ones(N, dtype=bool)          # built in place: memory is tight here
for k in SPECTRUM_KEYS:
    no_tachyon &= msq[k] > 0
mass = {k: np.sqrt(np.clip(v, 0.0, None)) for k, v in msq.items()}
del msq                                      # ~130 MB, fully consumed above

cuts = {}
cuts["sampled"]         = np.ones(N, dtype=bool)
cuts["Eq.(4) bounded"]  = bfb_mask(lam_scan)
cuts["+ corrected BFB"] = cuts["Eq.(4) bounded"] & neutral_real_bfb_mask(lam_scan)
cuts["+ unitarity"]     = cuts["+ corrected BFB"] & unitarity_mask(lam_scan)
cuts["+ no tachyons"]   = cuts["+ unitarity"] & no_tachyon

# LEP charged-Higgs bound: the 2HDM Type-II benchmark [LEPHpm13], applied as in [GomezBock21]
cuts["+ charged > 80"] = cuts["+ no tachyons"] & (mass["Hpm1"] > 80) & (mass["Hpm2"] > 80)

near125 = np.any(np.stack([np.abs(mass[k] - M_H) <= M_H_WINDOW
                           for k in ("h0", "H1", "H2")]), axis=0)
mass_only_125 = cuts["+ charged > 80"] & near125   # NOT a cut-flow row: the comparison of §8

# The coupling cut (section 8): the CP-even state CARRYING g_hVV must sit at M_H +- 3 GeV
# with kappa^2 >= KAPPA2_MIN, and no light state may be appreciably coupled ([LEP2003]).
# It chains from "+ charged > 80"; the carrier condition implies a state at M_H +- 3, which
# the assert below pins, so it is a subset of the mass-only selection.
hvv = hvv_function(m)
prev_idx = np.flatnonzero(cuts["+ charged > 80"])
xi_p = hvv([lam_scan[prev_idx, k] for k in range(8)], v2_s[prev_idx], vS_s[prev_idx])
m_p = {k: mass[k][prev_idx] for k in ("h0", "H1", "H2")}

def on_sample(mask_on_prev):
    full = np.zeros(N, dtype=bool)
    full[prev_idx[mask_on_prev]] = True
    return full

cuts["+ SM-like at 125"] = cuts["+ charged > 80"] & on_sample(sm_like_cut(xi_p, m_p))
cuts["+ LEP neutral veto"] = cuts["+ SM-like at 125"] & on_sample(lep_neutral_mask(xi_p, m_p))
assert not (cuts["+ SM-like at 125"] & ~mass_only_125).any()

prev = N
print("%-20s %9s %12s %12s" % ("cut", "surviving", "of sample", "of previous"))
for label, msk in cuts.items():
    n = int(msk.sum())
    print("%-20s %9d %11.3f%% %11.1f%%" % (label, n, 100 * n / N, 100 * n / prev))
    prev = max(n, 1)
print("\n(a mass-only M_H +- 3 GeV cut would instead keep %d)" % int(mass_only_125.sum()))
```

The cut flow is a random sample, so its percentages need an error bar. Splitting it in half
and comparing is the cheapest meaningful check:


```python
half = N // 2
print("%-20s %10s %10s %10s" % ("cut", "half A", "half B", "|diff|/sigma"))
for label, msk in cuts.items():
    pa, pb = msk[:half].mean(), msk[half:].mean()
    sigma = math.sqrt(max(pa * (1 - pa), 1e-12) / half) * math.sqrt(2)
    z = abs(pa - pb) / sigma if sigma > 0 else 0.0
    print("%-20s %9.4f%% %9.4f%% %10.2f" % (label, 100 * pa, 100 * pb, z))
    assert z < 5.0
ok("the two halves of the sample agree within Poisson fluctuations")
```

## 8. Benchmark points with a 125 GeV scalar

The measured 125 GeV boson is the state that couples to $W/Z$, so the cut here is on the
**coupling**, not on a mass coincidence. No gauge sector is needed for that: the $hVV$
coupling of every CP-even state is fixed by the same CP-even block the masses come from
(§8.1), and the cut is applied in the cut flow of §6 (§8.2 reads it back). Only the *labels*
of [GomezBock21]'s alignment scenarios A and B need the electroweak kinetic sector;
[`03_scalar_decays.ipynb`](03_scalar_decays.ipynb) adds it and recovers both as the
$\delta\to0$ and $\delta\to\pi/2$ limits.

### 8.1 Which CP-even state is the 125 GeV one?

A mass cut can only say that *some* CP-even state sits at $125.20\pm3$ GeV; it cannot say that
state is the one that couples to $W/Z$, which is what the measured 125 GeV boson does. Two facts make that checkable without leaving this
notebook. $h_0$ is gauge-phobic exactly, so $g^2_{h_0VV}=0$. The other two states share the
SM strength between them, $g^2_{H_1VV}+g^2_{H_2VV}=(g^{\rm SM})^2$, and each one's share is
the overlap of its eigenvector with the vacuum direction — index 0 of the geometric basis,
where the block $\begin{pmatrix}a&b\\b&c\end{pmatrix}$ lives:

$$g^2_{\pm}=\tfrac12\Big(1\pm\frac{a-c}{\sqrt{(a-c)^2+4b^2}}\Big),\qquad
\text{for the eigenvalue } \tfrac12\big(a+c\pm\sqrt{(a-c)^2+4b^2}\big).$$

**Check before trusting it.** Diagonalize the *un-rotated* $3\times3$ CP-even matrix
numerically and project each eigenvector on $\hat v=(v_1,v_2,v_S)/v$; that is the coupling
by definition, with no geometric-basis assumption in it. *Would be falsified by* any point
where the two disagree, or where the three overlaps fail to sum to 1.

Once validated, this is what §8.2 cuts on.


```python
M_S = mass_matrices(m)[0]
hvv = hvv_function(m)

# validate on a random subset of the points that REACH the coupling cut -- a
# stronger test than checking only the few that survive it, and the subset keeps
# the per-point symbolic diagonalization affordable.
pool = np.flatnonzero(cuts["+ charged > 80"])
probe = np.random.default_rng(7).choice(pool, size=min(300, len(pool)), replace=False)
g2_probe = hvv([lam_scan[probe, k] for k in range(8)], v2_s[probe], vS_s[probe])

t0 = time.time()
worst_sum = worst_proj = 0.0
for j, i in enumerate(probe):
    point = {**dict(zip(m.lam_symbols, lam_scan[i])), m.v2.s: v2_s[i], m.vS.s: vS_s[i]}
    Mn = np.array(sp.N(M_S.subs(point)).tolist(), dtype=float)
    w, U = np.linalg.eigh((Mn + Mn.T) / 2)
    vhat = np.array([v12_s[i] / 2, v2_s[i], vS_s[i]])
    vhat = vhat / np.linalg.norm(vhat)
    proj = np.sort((U.T @ vhat) ** 2)                 # the coupling by definition
    ours = np.sort([g2_probe[k][j] for k in ("h0", "H1", "H2")])
    worst_proj = max(worst_proj, np.abs(proj - ours).max())
    worst_sum = max(worst_sum, abs(sum(g2_probe[k][j] for k in ("h0", "H1", "H2")) - 1.0))
print("checked %d of the %d points reaching the coupling cut, in %.1f s"
      % (len(probe), len(pool), time.time() - t0))
print("max |block overlap - eigenvector projection| : %.1e" % worst_proj)
print("max |sum of the three g^2 - 1|                : %.1e" % worst_sum)
assert worst_proj < 1e-9 and worst_sum < 1e-12
assert (g2_probe["h0"] == 0.0).all()
ok("the block overlaps ARE the eigenvector projections on the vacuum direction; h0 is "
   "exactly gauge-phobic and the three sum to the SM strength")

ledger.step(r"The $hVV$ couplings are the CP-even block overlaps with the vacuum direction; "
            r"$h_0$ is exactly gauge-phobic",
            obtained="derived",
            checks=("block overlaps equal the eigenvector projections of the un-rotated 3x3 "
                    "matrix at %d points" % len(probe),
                    r"the three $g^2$ sum to 1 to $10^{-12}$",
                    r"$g^2_{h_0VV}=0$ exactly"),
            falsified_by="a point where overlap and projection disagree, or the sum is not 1",
            section="8.1")
```

### 8.2 Cutting on the coupling

With $\xi^2$ in hand the 125 GeV condition becomes a statement about the state that
*carries* the coupling, and the light neutral states become constrainable. Two thresholds,
both named in `constraints.py`: $\kappa^2\ge0.9$ for the carrier, and $\xi^2<0.05$ for
anything below 114.4 GeV ([LEP2003] Fig. 10, as a step).

The second looks like it needs a digitized exclusion curve. It does not, and the table
below is the reason: the two non-gauge-phobic states share the SM strength exactly, so
$\kappa^2\ge0.9$ already forces the other one under $0.1$ — the LEP floor then removes
nothing, for any value across an order of magnitude. *Would be falsified by* the floor
changing the count at the adopted $\kappa^2$. Both thresholds are applied in §6's cut flow;
the tables below show how much each one does.

```python
# How much each threshold does, on the points that reached them.  The two
# non-phobic states share the SM strength, so kappa^2 >= 0.9 already forces the
# other under 0.1 -- which is why the LEP floor removes nothing here (see
# constraints.XI2_LEP_FLOOR) and the result does not rest on approximating its curve.
# xi_p, m_p: the coupling and masses of every point reaching the cut, from section 6

print("%-14s %10s %10s %s" % ("kappa^2 >=", "at 125", "+ LEP veto", "lost to veto"))
for thr in (0.5, 0.8, KAPPA2_MIN, 0.95, 0.99):
    a = sm_like_cut(xi_p, m_p, kappa2_min=thr)
    b = a & lep_neutral_mask(xi_p, m_p)
    print("%-14.2f %10d %10d %d" % (thr, a.sum(), b.sum(), a.sum() - b.sum()))

base = sm_like_cut(xi_p, m_p)
print("\nat the adopted kappa^2 = %.2f, the LEP floor changes nothing over:" % KAPPA2_MIN)
print("   ", {f"{f:.2f}": int((base & lep_neutral_mask(xi_p, m_p, floor=f)).sum())
              for f in (0.01, 0.02, 0.05, 0.10)})
assert len({int((base & lep_neutral_mask(xi_p, m_p, floor=f)).sum())
            for f in (0.01, 0.02, 0.05, 0.10)}) == 1
ok("the surviving set is independent of the LEP floor across an order of magnitude, "
   "so the step approximation of [LEP2003] Fig. 10 carries no weight in the result")
```

```python
# The coupling cut is applied in the cut flow of section 6; here it is read back.
idx = np.flatnonzero(cuts["+ LEP neutral veto"])
print("thresholds: kappa^2 >= %.2f, and xi^2 < %.2f below %.1f GeV"
      % (KAPPA2_MIN, XI2_LEP_FLOOR, M_LEP_ZH))
print("\n%-22s %9s" % ("cut", "surviving"))
for label in ("+ charged > 80", "+ SM-like at 125", "+ LEP neutral veto"):
    print("%-22s %9d" % (label, int(cuts[label].sum())))
print("\nmass-only 'M_H +- 3 GeV' would have kept %d" % int(mass_only_125.sum()))
ok("the 125 GeV cut is now a COUPLING condition, not a mass coincidence")
check_quoted("8.2, §9 and the report", [
    ("mass-only M_H +- 3 survivors", int(mass_only_125.sum()), 313, 0),
    ("coupling-cut survivors", int(cuts["+ LEP neutral veto"].sum()), 64, 0)])
```

## 9. Findings

1. **The tutorial's benchmark vacuum sits at $v=155$ GeV, not 246.** Imposing the √3
   alignment as a substitution silently overwrites $v_1$ and breaks the scale constraint
   that motivated the original numbers. Every mass quoted in its §9 is low by a factor
   1.587. The fix is to impose both conditions at once (`aligned_vevs`), leaving θ as the
   one free vacuum parameter. The alignment itself is not an input: it is forced by the
   over-constrained tadpole system, whose residual $\tfrac32\lambda_4v_S(3v_1^2-v_2^2)$
   vanishes only on $v_2=\sqrt3v_1$ (§2.1).

2. **The tutorial's benchmark λ's are not a viable theory** — the potential is not bounded
   from below there. "5 of 7 states are tachyonic" was a symptom; the point was never
   admissible.

3. **feynlag's λ basis is *not* [DasDey14]'s, and the difference is exactly one sign.**
   The λ₄ invariant transcribed literally from the paper fails S₃ invariance in feynlag's
   real-orthogonal irrep basis (§4). Seven of the eight quartic structures are invariant
   under *any* $SO(2)$ rotation of the doublet, so their coefficients transfer directly;
   only $x_S\cdot d_2$ can change, and the sole S₃-preserving transformation that changes it
   is $H_{1,2}\to-H_{1,2}$, which sends $\lambda_4\to-\lambda_4$. Matching feynlag's masses
   to [GomezBock21] Eqs. (30)–(33) fixes the sign ($e=-\lambda_4$) and validates the whole
   chain against published closed forms. Because the sign flip is a field redefinition, no
   physical condition may depend on it — and indeed λ₄ enters the constraints only as
   $|\lambda_4|$ or $\lambda_4^2$, so they transfer unchanged.

4. **The published boundedness conditions are not sufficient** (§5.3–§5.5). Roughly **10%**
   of the λ points that [DasDey14] Eq. (4) accepts have an explicit real neutral direction
   along which $V_4<0$. The cause is exact and simple: on that slice boundedness requires
   $f(t)=\lambda_8t^4+(\lambda_5+\lambda_6+2\lambda_7)t^2-2|\lambda_4|t+(\lambda_1+\lambda_3)\ge0$
   for **all** $t\ge0$, and Eq. (4g) is precisely $f(1)>0$ — one point of that curve. The
   corrected condition is derived, implemented (`constraints.neutral_real_bfb_min`) and
   validated against brute-force sampling. The 2015 erratum does **not** fix it — it
   corrects only the tadpole equations (§2.2) — so this is an insufficiency of the final
   published conditions, not of a superseded pre-erratum form.

5. **Boundedness, not unitarity, is the binding constraint** in this λ range. Tree unitarity
   ($|a|\le16\pi$) is essentially never violated for $|\lambda|\le2$ — it would only bite at
   $|\lambda|\gtrsim$ several — while boundedness removes the overwhelming majority of
   points, and the corrected condition removes another ~10% on top of Eq. (4).

6. **The viable spectra cluster below a TeV**, consistent with [DasDey14]'s own conclusion
   that "many new scalars must be lurking below 1 TeV".

7. **The 125 GeV condition is a coupling cut, and it is five times tighter than a mass cut**
   (§8). Requiring the state that carries $g_{hVV}$ to sit at $125.20\pm3$ GeV ($M_H$, PDG 2024) with
   $\kappa^2\ge0.9$ keeps 64 of the 2M points, against 313 for a mass-only cut; the LEP
   neutral veto and the 12-d boundedness backstop then remove nothing. Over the 30 chunks of
   `scan.py`, **1973 of $6\times10^7$** sampled points survive (`results/viable_points.json`),
   and chunk 0 reproduces this notebook's sample point for point.
